````python
import cufflinks as cf
from plotly.offline import iplot
cf.go_offline()
````

### Plotando candlesticks e médias móveis simples e exponenciais

````python
cs_bitcoin = cf.QuantFig(
    bitcoin, title="Bitcoin cotation",
    legend="top", name="Bitcoin cotation"
    )

cs_bitcoin.add_volume()
cs_bitcoin.add_sma(periods=40, column="Close", color="red")
cs_bitcoin.add_ema(periods=40, color="green")

cs_bitcoin.iplot()
````
